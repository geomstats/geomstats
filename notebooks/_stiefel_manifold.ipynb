## Introduction and definition

The Stiefel manifold $\operatorname{St}(n, p), n \geq p$, is a Riemannian manifold that is composed of all $n \times p$ orthonormal matrices:

$$
\operatorname{St}(n, p) = \left\{X \in \mathbb{R}^{n \times p}: X^{T} X=I_{p}\right\}
$$

It has dimension equal to $np-\frac{1}{2}p(p+1) .$ We will regard this manifold as an embedded sub-maniford of $\mathbb{R}^{n \times p} .$ This means that we identify tangent vectors to the manifold
with $n \times p$ matrices. The tangent space of $\operatorname{St}(n, p)$ at $X$ is:

$$
T_{X}\operatorname{St}(n, p)=\left\{Z \in \mathbb{R}^{n \times p}: Z^{T} X+X^{T} Z=0\right\}
$$

Note than if $p = 1$, the Stiefel manifold is a Hypersphere of dimension $n-1$:
$$
\operatorname{St}(n,1) = \left\{x \in \mathbb{R}^{n}: \|x\| =1 \right\}
$$

If $p = n$, the Stiefel manifold is a Orthogonal group of dimension $\frac{1}{2}n^{2}-\frac{1}{2}n$:
$$
\operatorname{St}(n,n) = \left\{X \in \mathbb{R}^{n \times n}: X^{T} X=I_{n} \right\}
$$

We have inplemented the visualization of three specific cases of Stiefel manifold: $\operatorname{St}(2, 1), \operatorname{St}(2, 2)$ and $\operatorname{St}(3, 1)$. In this presentation, we mainly focus on the space $\operatorname{St}(2, 2)$ of $2 \times 2$ orthogonal matrices. This space is especially Interesting to look at because it has dimension 1 and there exists a projection such that we can directly visualize $2 \times 2$ orthogonal matrices on a circle.
This rest of the presentation is organized as follows:
- First we study the theory of the Stiefel manifold $\operatorname{St}(2, 2)$.
- Then we present the visualization of the Stiefel manifolds.
- Finally we show some real-world applications.

## Theory: Stiefel manifold $\operatorname{St}(2, 2)$

Let $$X =\begin{bmatrix} x_{1} & x_{2} \\ y_{1} & y_{2} \end{bmatrix}$$ 

be an element of $\operatorname{St}(2, 2)$. Then we have $x_{1}^{2} + y_{1}^{2} = 1, x_{2}^{2} + y_{2}^{2} = 1$ and $x_{1}x_{2} + y_{1}y_{2}=0$. This means $[x_{1}, y_{1}]^{T}$ and $[x_{2}, y_{2}]^{T}$ are two orthogonal unit vectors so that we can just use one of them to represent the matrix $X.$ The question is: which vector to choose, the first column or the second column of $X$? 

For convenience of our visualization, we use the following way to represent $X$ on a unit circle using just a point $x$ on the circle:

$$
x = \begin{cases} [x_{1}, y_{1}]^{T}, & \text{if} \operatorname{det}(X) > 0 \\ [x_{2}, y_{2}]^{T},  &  \text{if} \operatorname{det}(X) < 0   \end{cases}
$$

Hence by rotating x through 90 degree about the origin in anticlockwise direction, we can get the other column of the matrix.

Now let $$Z =\begin{bmatrix} x_{3} & x_{4} \\ y_{3} & y_{4} \end{bmatrix}$$ 

be an tangent vector of the tangent space $T_{X}\operatorname{St}(2, 2) .$ Then we have $x_{3}x_{1} + y_{3}y_{1} = 0, x_{4}x_{2} + y_{4}y_{2} = 0$ and $x_{3}x_{2} + y_{3}y_{2} = -x_{4}x_{1}-y_{4}y_{1}$. This means $[x_{3}, y_{3}]^{T}$ and $[x_{4}, y_{4}]^{T}$ are on the tangent lines defined by $[x_{1}, y_{1}]^{T}$ and $[x_{2}, y_{2}]^{T}$ correspondingly. Then we note that:

$$
x_{3}x_{2} + y_{3}y_{2} = \begin{cases} |[x_{3}, y_{3}]^{T}|, & \text{if} [x_{3}, y_{3}]^{T} \text{is in the same direction as}~[x_{2}, y_{2}]^{T} \\ -|[x_{3}, y_{3}]^{T}|, & \text{otherwise}  \end{cases}
$$

Similarly:

$$
x_{4}x_{1} + y_{4}y_{1} = \begin{cases} |[x_{4}, y_{4}]^{T}|, & \text{if} [x_{4}, y_{4}]^{T} \text{is in the same direction as}~[x_{1}, y_{1}]^{T} \\ -|[x_{4}, y_{4}]^{T}|, & \text{otherwise}  \end{cases}
$$

Hence $|[x_{3}, y_{3}]^{T}| = |[x_{4}, y_{4}]^{T}|.$ If $[x_{3}, y_{3}]^{T}$ is in the same direction as $[x_{2}, y_{2}]^{T}$, $[x_{4}, y_{4}]^{T}$ has to be in the opposite direction as $[x_{1}, y_{1}]^{T}$ and vice versa.

```python
### import some packages
# begin
import matplotlib.pyplot as plt
from math import cos, sin
import numpy as np

import geomstats.backend as gs
import geomstats.datasets.utils as data_utils
from geomstats.visualization.stiefel_manifold import \
    StiefelSphere, StiefelCircle, Arrow2D
from geomstats.geometry.stiefel import Stiefel, StiefelCanonicalMetric
np.random.seed(seed=15)

# end
```

## Visualization of Stiefel manifold $\operatorname{St}(2, 1)$: Regularly sampled data and tangent space

```python
###########################################################################
#### Consider <n=2, p=1> Stiefel Manifold.
#### Dimension is 1.
###########################################################################
# %%

### Define the manifold space and circle instance.
# begin
v21 = Stiefel(2,1)
fig = plt.figure(figsize=(12,12))
circle = StiefelCircle()
ax = circle.set_ax(ax=None)
# end

### Plot the regular sampled points and tangent space
# begin
points = v21.random_uniform(100)[:,:,0]
circle.add_points(points)
circle.draw(ax, markersize = 5, alpha = 0.5, label = 'points')
circle.draw_tangent_space(ax, points[0], label = 'tangent space', color = 'g', alpha = 0.5)
ax.legend(prop={'size': 12})
plt.title('n=2, p=1 Manifold', size = 15)
# end
```

## Visualization of Stiefel manifold $\operatorname{St}(2, 2)$: Regularly sampled points

```python
###########################################################################
#### Consider <n=2, p=2> Stiefel Manifold.
#### Dimension is 1, which is two related vectors and
#### can be represented as one point in the circle.
###########################################################################

# %%
### Define the manifold space and corresponding metric.
# begin
v22 = Stiefel(2,2)
metric_v22 = StiefelCanonicalMetric(2,2)
# end

# %%
### Generate the points and rearrange them as proper order.
### And we choose one matrix to show our tangent space.

# begin
points = v22.random_uniform(100)
det = np.linalg.det(points)
points_to_circle = np.zeros((100,2))

for i in range(100):
    if det[i] > 0:
        points_to_circle[i,:] = points[i][:,0]
    elif det[i] < 0:
        points_to_circle[i,:] = points[i][:,1]

if det[0] > 0:
    plot_base_point = points[0][:,0]
    plot_respon_point = points[0][:,1]
else:
    plot_base_point = points[0][:,1]
    plot_respon_point = points[0][:,0]
# end 

# %%
### Plot the regular sampled points.
# begin
fig = plt.figure(figsize=(12,12))
circle = StiefelCircle()
ax = circle.set_ax(ax=None)
circle.draw(ax, markersize = 30, alpha = 0.5)
circle.add_points(points_to_circle)
circle.draw_points(ax, markersize = 5, c = 'b', alpha = 0.5)
circle.clear_points()
plt.title('<n=2,p=2> regular sampled points', size = 15)
# end
```

## Visualization of Stiefel manifold $\operatorname{St}(2, 2)$: Tangent space and tangent vector

```python
### Plot the tangent space.
# begin
fig = plt.figure(figsize=(12,12))
circle = StiefelCircle()
ax = circle.set_ax(ax=None)
circle.add_points(np.expand_dims(plot_base_point, axis=0))
circle.draw(ax, markersize = 30, alpha = 0.5)
circle.draw_tangent_space(ax, plot_base_point, label = 'tangent space', color = 'g', alpha = 0.5)
# circle.draw_tangent_space(ax, plot_respon_point, label = 'tangent line', color = 'red', alpha = 0.5)

# arrow_from_origin right
arrow_right = Arrow2D([0,0], vector=plot_base_point); 
arrow_right.draw(ax, color="black", scale=5, label='original vector'); 

# arrow_from_origin left
arrow_left = Arrow2D([0,0], vector=plot_respon_point); 
arrow_left.draw(ax, color="orange", scale=5, label='original vector'); 

theta = np.deg2rad(90)
rot = np.array([[cos(theta), -sin(theta)], [sin(theta), cos(theta)]])
v = np.dot(rot, plot_base_point)
# arrow_tangent right
arrow_right = Arrow2D(plot_base_point, vector=v); 
arrow_right.draw(ax, color="brown", scale=8, label='tangent vector'); 

theta = theta
rot = np.array([[cos(theta), -sin(theta)], [sin(theta), cos(theta)]])
v = np.dot(rot, plot_respon_point)
# arrow_tangent left
arrow_left = Arrow2D(plot_respon_point, vector=v); 
arrow_left.draw(ax, color="blue", scale=8, label='tangent vector'); 
ax.legend(prop={'size': 12});
plt.title('<n=2,p=2> Stiefold Manifold Representation and Tangent Space', size = 15)
# end
```

## Visualization of Stiefel manifold $\operatorname{St}(2, 2)$: Exp operation

```python
### Try some operation to this manifold space.
# begin
N = 2

group1_points = v22.random_uniform(N)
group2_points = v22.random_uniform(N)

det = np.linalg.det(group1_points)
group1_points_to_circle = np.zeros((N,2,2))
for i in range(N):
    if det[i] > 0:
        group1_points_to_circle[i,:,0] = group1_points[i][:,0]
        group1_points_to_circle[i,:,1] = group1_points[i][:,1]
    elif det[i] < 0:
        group1_points_to_circle[i,:,0] = group1_points[i][:,1]
        group1_points_to_circle[i,:,1] = group1_points[i][:,0]
det = np.linalg.det(group2_points)
group2_points_to_circle = np.zeros((N,2,2))
for i in range(N):
    if det[i] > 0:
        group2_points_to_circle[i,:,0] = group2_points[i][:,0]
        group2_points_to_circle[i,:,1] = group2_points[i][:,1]
    elif det[i] < 0:
        group2_points_to_circle[i,:,0] = group2_points[i][:,1]
        group2_points_to_circle[i,:,1] = group2_points[i][:,0]


theta = np.deg2rad(90)
rot = np.array([[cos(theta), -sin(theta)], [sin(theta), cos(theta)]])
group1_tangent_v = np.zeros((N,2,2))
for i in range(group1_points_to_circle.shape[0]):
    group1_tangent_v[i] = np.dot(rot, group1_points_to_circle[i])

group1_tangent = v22.to_tangent(group1_tangent_v[:,:,:], group1_points_to_circle)
group1_exp = metric_v22.exp(group1_tangent, group1_points_to_circle)
# end 

# %%
### Try some operation to this manifold space.
# begin
det = np.linalg.det(group1_exp)
group1_exp_points_to_circle = np.zeros((N,2,2))
for i in range(N):
    if det[i] > 0:
        group1_exp_points_to_circle[i,:,0] = group1_exp[i][:,0]
        group1_exp_points_to_circle[i,:,1] = group1_exp[i][:,1]
    elif det[i] < 0:
        group1_exp_points_to_circle[i,:,0] = group1_exp[i][:,1]
        group1_exp_points_to_circle[i,:,1] = group1_exp[i][:,0]

fig = plt.figure(figsize=(12,12))
circle = StiefelCircle()
ax = circle.set_ax(ax=None)
circle.add_points(group1_exp_points_to_circle[:,:,0])
circle.draw(ax, markersize = 10, c = 'r', label = 'points after exp operation')
circle.clear_points()
circle.add_points(group1_points_to_circle[:,:,0])
circle.draw(ax, markersize = 10, c = 'b', label = 'original ')
circle.clear_points()
theta = np.deg2rad(90)
rot = np.array([[cos(theta), -sin(theta)], [sin(theta), cos(theta)]])
v = np.dot(rot, group1_points_to_circle[0,:,0])
# arrow_tangent right
arrow_right = Arrow2D(group1_points_to_circle[0,:,0], vector=v); 
arrow_right.draw(ax, color="blue", scale=8, label='tangent vector');

theta = theta
rot = np.array([[cos(theta), -sin(theta)], [sin(theta), cos(theta)]])
v = np.dot(rot, group1_points_to_circle[1,:,0])
# arrow_tangent left
arrow_left = Arrow2D(group1_points_to_circle[1,:,0], vector=v); 
arrow_left.draw(ax, color="blue", scale=8, label='tangent vector');
ax.legend(prop={'size': 12})
plt.title('Exp-transformed points comparison with original points ', size = 15)
```

## Applications

Stiefel manifolds represent the orthogonality constraints that arise in such areas as the symmetric eigenvalue problem, nonlinear eigenvalue problems, signal processing and machine learning [1].

### Orthogonal weight normalization

Strictly enforcing orthonormality constraints on parameter matrices has been shown advantageous in deep learning. In [2], they proposed a novel orthogonal weight normalization method to solve optimization over Stiefel manifolds.

### Soft Orthogonality regularization

Soft Orthogonality (SO) regularization [3] was proposed to relax the “hard orthogonality” assumption [2], which can reduce the training cost:
$$
(\mathrm{SO}) \quad \lambda\left\|W^{T} W-I\right\|_{F}^{2}
$$

### Principal component analysis

Let $A=A^{T} \in \mathbb{R}^{n \times n}$.

Goal: Compute the $p$ dominant eigenvectors of $A$.

Principle: Let $N=\operatorname{diag}(p, p-1, \cdots, 1)$ and solve
$$
\max _{X^{T} X=I_{p}} \operatorname{tr}\left(X^{T} A X N\right)
$$
The columns of $X$ are the $p$ dominant eigenvectors or $A$. 

A basic method: Steepest-descent on $\operatorname{St}(n, p)$. We refer readers to [] to see more details of this application

###  Procrustes problem on the Stiefel manifold

In [4], they stuied the orthogonal procrustes problem: given $A \in \mathbb{R}^{m \times n}, B \in \mathbb{R}^{m \times p}$. 

Goal: Find $Q \in \mathbb{R}^{n \times p}$ that solves
$$
\min _{Q^{T} Q=I_{p}}\|A Q-B\|_{F}^{2}
$$

Applications: factor analysis, used notably in psychometrics.

### Blind source separation

 Joint diagonalization on the Stiefel manifold [6] can be applied to do blind source separation, the Two mixed pictures are given as input to a blind source separation algorithm based on a trust-region method on $\operatorname{St}(2, 2)$.
